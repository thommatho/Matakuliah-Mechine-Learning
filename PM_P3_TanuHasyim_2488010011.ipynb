## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
# TULIS KODE ANDA DI SINI (Latihan 1)
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='darkgrid')
df = sns.load_dataset('tips')
print(df.shape)
df.info()
df.head()
```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
# TULIS KODE ANDA DI SINI (Latihan 2)
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

plt.figure(figsize=(8, 5))
sns.histplot(data=tips, x='total_bill', bins=15)
plt.title('Histogram Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')
plt.show()

plt.figure(figsize=(8, 4))
sns.boxplot(data=tips, x='tip')
plt.title('Box Plot Tip')
plt.xlabel('Tip')
plt.show()
```

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# TULIS KODE ANDA DI SINI (Latihan 3)

import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time'
)

plt.title('Total Bill vs Tip berdasarkan Waktu')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?

```python
# TULIS KODE ANDA DI SINI (Latihan 4)
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

num = tips.select_dtypes(include='number')

plt.figure(figsize=(8, 6))
sns.heatmap(
    num.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Heatmap Korelasi Dataset Tips')
plt.show()
```
