1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?

Supervised Learning adalah pembelajaran menggunakan data yang sudah memiliki label. Model belajar dari hubungan antara fitur dan label untuk melakukan prediksi. Contohnya memprediksi harga rumah atau menentukan apakah siswa lulus.
Unsupervised Learning menggunakan data tanpa label. Model berusaha menemukan pola atau struktur yang terdapat di dalam data. Contohnya mengelompokkan pelanggan berdasarkan karakteristik tertentu.
Reinforcement Learning adalah pembelajaran melalui interaksi dengan lingkungan. Model atau agen akan mendapatkan reward ketika melakukan tindakan yang benar dan penalty ketika melakukan tindakan yang kurang tepat. Contohnya pembelajaran agen untuk bermain game.

2. Mengapa pembagian data latih dan uji penting?

Pembagian data latih dan uji penting untuk mengetahui apakah model mampu melakukan generalisasi terhadap data yang belum pernah dilihat sebelumnya. Data latih digunakan untuk mempelajari pola, sedangkan data uji digunakan untuk mengevaluasi performa model. Dengan demikian, kita dapat mengetahui apakah model benar-benar belajar pola dari data atau hanya menghafal data pelatihan (overfitting).

3. Mengapa alur kerja ML disebut iteratif?

Alur kerja Machine Learning disebut iteratif karena prosesnya dilakukan berulang kali. Setelah model dilatih dan dievaluasi, hasil evaluasi digunakan untuk melakukan perbaikan, seperti mengubah fitur, memilih algoritma lain, mengatur parameter, atau menambah data. Proses pelatihan → evaluasi → perbaikan kemudian diulangi sampai diperoleh model dengan performa yang sesuai.
