Langkah 1. Mengatasi Nilai Hilang



```python
import pandas as pd
import numpy as np

# Dataset Tanaman
data = {
    'tinggi_cm'       : [15.2, 22.5, np.nan, 31.4, 18.7,
                         40.2, 27.8, np.nan, 20.1, 35.6],

    'jumlah_daun'     : [8, 12, 7, np.nan, 10,
                         18, 14, 9, np.nan, 16],

    'ph_tanah'        : [6.2, 6.8, 5.9, 7.1, np.nan,
                         6.5, 6.9, 5.8, 6.3, np.nan],

    'jenis_tanaman'   : ['Mawar', 'Melati', 'Mawar', 'Anggrek', 'Melati',
                         'Anggrek', 'Mawar', 'Melati', 'Mawar', 'Anggrek'],

    'intensitas_cahaya': ['Rendah', 'Sedang', 'Rendah', 'Tinggi', 'Sedang',
                          'Tinggi', 'Sedang', 'Rendah', 'Tinggi', 'Sedang'],

    'media_tanam'     : ['Tanah', 'Tanah', 'Hidroponik', 'Tanah', 'Hidroponik',
                         'Tanah', 'Hidroponik', 'Tanah', 'Hidroponik', 'Tanah'],

    'sehat'           : ['Tidak', 'Ya', 'Tidak', 'Ya', 'Ya',
                         'Ya', 'Ya', 'Tidak', 'Tidak', 'Ya']
}
df = pd.DataFrame(data)
df['tinggi_cm'] = df['tinggi_cm'].fillna(df['tinggi_cm'].median())
df['jumlah_daun'] = df['jumlah_daun'].fillna(df['jumlah_daun'].median())
df['ph_tanah'] = df['ph_tanah'].fillna(df['ph_tanah'].median())

print(df)
```

Langkah 2. Memisahkan Fitur (X) dan Label (Y)

```python
X = df.drop(columns=['sehat'])
y = df['sehat']
print(f"Tabel data Fitur\n{X}\n")
print(f"Tabel data Label\n{y}")
```

Langkah 3. Encoding Kategorical (Ordinal & one-hot)

```python
X['intensitas_cahaya'] = X['intensitas_cahaya'].map({'Rendah':0,'Sedang':1,'Tinggi':2})
X['media_tanam'] = X['media_tanam'].map({'Tanah':0, 'Hidroponik':1}) # ordinal
X = pd.get_dummies(X, columns=['jenis_tanaman'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
print(f"{X}\n{y}")
```

Langkah 4. Membagi data (Sebelum Scaling)

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

Langkah 5. Penskalaan (fit pada latih, transform pada latih & uji)

```python
from sklearn.preprocessing import StandardScaler

num = ['tinggi_cm', 'jumlah_daun', 'ph_tanah']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])
```

Langkah 6. Verifikasi

```python
print(f"Periksa nilai yang masih hilang{X_train.isnull().sum()}\n")
print(f"{X_test.isnull().sum()}\n")
print(f"periksa tipe data{X_train.dtypes}\n")
print(f"Periksa hasil scaling{X_train[num].describe()}")
```

4. Jelaskan mengapa fit_transform pada latih tetapi hanya pada transform pada uji?

fit_transform() digunakan pada data latih karena scaler perlu mempelajari parameter dari data tersebut, seperti mean dan standar deviasi pada StandardScaler.
transform() digunakan pada data uji karena data uji harus diproses menggunakan parameter yang telah dipelajari dari data latih.

1. Mengapa Urutan split dulu, baru scaling penting?

Urutan tersebut penting untuk mencegah data leakage, yaitu ketika informasi dari data uji ikut memengaruhi proses pelatihan.

Jika scaling dilakukan sebelum pembagian data, parameter scaling dapat dihitung menggunakan seluruh dataset, termasuk data uji. Akibatnya, informasi dari data uji dapat masuk ke proses preprocessing data latih.

2. Kapan label encoding dan kapan one-hot encoding

Label encoding digunakan ketika kategori memiliki hubungan urutan atau ketika kategori biner perlu diubah menjadi angka.

One-hot encoding digunakan untuk kategori nominal yang tidak memiliki tingkatan.

3. Apa perbedaan normalisasi dan standardisasi?

Normalisasi mengubah nilai fitur ke rentang tertentu, misalnya 0 sampai 1.

Standardisasi mengubah nilai fitur sehingga memiliki rata-rata mendekati 0 dan standar deviasi mendekati 1.

Keduanya bertujuan membuat skala fitur lebih seragam, tetapi cara perhitungannya berbeda.

| Aspek             | Normalisasi                  | Standardisasi                      |
| ----------------- | ---------------------------- | ---------------------------------- |
| Teknik umum       | Min-Max Scaling              | Standard Scaling                   |
| Rentang hasil     | Biasanya 0–1                 | Tidak memiliki batas tetap         |
| Parameter         | Minimum dan maksimum         | Mean dan standar deviasi           |
| Tujuan            | Menyamakan rentang fitur     | Memusatkan data pada mean 0        |
| Contoh penggunaan | Fitur dengan rentang berbeda | Model yang sensitif terhadap skala |
